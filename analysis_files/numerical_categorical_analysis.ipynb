## Numerical Analysis and Visualization

```python
#analyzing a specific column
df['Booking_Value'].describe()
#prints column summary statistics
```

```python
df['Booking_Value'].median()

#median of column
```

```python
df['Booking_Value'].plot(kind='box', vert=False, figsize=(14, 6))
#box plot of column
```

```python
df['Booking_Value'].plot(kind='density', figsize=(14, 6))
#density plot of column
```

```python
ax = df['Booking_Value'].plot(kind='density', figsize = (14,6))
ax.axvline(df['Booking_Value'].mean(), color='red')
ax.axvline(df['Booking_Value'].median(), color='green')
#density plot with mean and median lines
```

```python
ax = df['Booking_Value'].plot(kind='hist',
                              bins=30,
                                color='skyblue', 
                                edgecolor = 'black',
                                figsize=(14,6))
#labels and title
ax.set_ylabel('Number of Bookings')
ax.set_xlabel('Booking Value')
ax.set_title('Distribution of Booking Values')
```

## Categorical Analysis and Visualzation

```python
df.head()
```

```python
df['Vehicle_Type'].value_counts()
#prints categories and count of categories in column 
```

```python
df['Vehicle_Type'].value_counts().plot(kind='pie', figsize=(8,8), autopct='%1.1f%%', startangle=140)
plt.title('Vehicle Types')

#prints pie chart of column "vehicle types" categories
```

```python
ax = df['Vehicle_Type'].value_counts().plot(kind='bar', color='pink', figsize=(14,6))
ax.set_xlabel('Types of Vehicles')
ax.set_ylabel('Number of Vehicle Types')
ax.set_title('Number of Vehicle Types driven for Uber')
#bar chart of column "vehicle types" categories
```
